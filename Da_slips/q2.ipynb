```python

import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.ensemble import RandomForestClassifier

df = pd.DataFrame({
    "Application_ID": range(1,11),
    "Age": [25,32,45,28,50,35,40,23,48,31],
    "Income": [30,50,80,35,90,60,75,25,85,45],
    "Credit_Score": [600,680,750,620,780,710,740,580,770,660],
    "Loan_Amount": [10,20,30,12,25,18,22,8,28,15],
    "Employment_Years": [2,5,15,3,20,8,12,1,18,4],
    "Debt_Ratio": [.45,.30,.20,.50,.15,.28,.22,.60,.18,.35],
    "Loan_Status": ["Rejected","Approved","Approved","Rejected",
                    "Approved","Approved","Approved","Rejected",
                    "Approved","Rejected"]
})

df.to_csv("Loan_application.csv", index=False)
X = df.drop(columns=["Application_ID","Loan_Status"])
y = df["Loan_Status"]

rf = RandomForestClassifier(random_state=42)
rf.fit(X,y)
print("Feature Importance:\n",
      pd.Series(rf.feature_importances_, index=X.columns))

X = StandardScaler().fit_transform(X)
pca = PCA(n_components=2)
print("PCA Data:\n", pca.fit_transform(X))
print("Variance:", pca.explained_variance_ratio_)
```
