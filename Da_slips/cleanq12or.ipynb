```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. Create Salary Dataset
df = pd.DataFrame({
    'Experience': [1.1, 1.3, 1.5, 2.0, 2.2, 3.0, 4.0, 5.0],
    'Salary': [39343, 46205, 37731, 43525, 39891, 60150, 63218, 83088]
})

# 2. Split into Train and Test sets
X = df['Experience'].values
y = df['Salary'].values
split = int(len(df) * 0.8)

X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# 3. Train Linear Regression Model (y = mx + c)
m, c = np.polyfit(X_train, y_train, 1)

# 4. Predict and Calculate Mean Squared Error (MSE)
y_pred = m * X_test + c
mse = np.mean((y_test - y_pred) ** 2)
print(f"Mean Squared Error (MSE): {mse:.2f}")

# 5. Plot Regression Line
plt.scatter(X_train, y_train, color='blue', label='Train Data')
plt.scatter(X_test, y_test, color='green', label='Test Data')
plt.plot(X_train, m * X_train + c, color='red', label='Regression Line')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.legend()
plt.show()
```
