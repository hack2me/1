```python

import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

df = pd.DataFrame({
    "Price": [500,700,450,900,650],
    "Rating": [4.2,4.5,3.8,4.8,4.1],
    "Sales": [100,150,80,200,120]
})

Z = linkage(df, method="ward")
dendrogram(Z, labels=["A","B","C","D","E"])
plt.title("Product Dendrogram")
plt.show()
```
